```python
fig, ax = plt.subplots(1,3, figsize=(9,3), sharey='row')

ax[0].plot(dataFpd["km"],dataFpd["avgPrice"], linestyle="none", marker=".")
ax[1].plot(dataFpd["year"],dataFpd["avgPrice"], linestyle="none", marker=".")
ax[2].plot(dataFpd["powerPS"],dataFpd["avgPrice"], linestyle="none", marker=".")

#ax[0].plot(dataFpd["avgPrice"],dataFpd["km"], linestyle="none", marker=".")
#ax[1].plot(dataFpd["avgPrice"],dataFpd["year"], linestyle="none", marker=".")
#ax[2].plot(dataFpd["avgPrice"],dataFpd["powerPS"], linestyle="none", marker=".")

ax[0].set_ylabel("Price")

ax[0].set_xlabel("km")
ax[1].set_xlabel("year")
ax[2].set_xlabel("power")

ax[0].grid(True)
ax[1].grid(True)
ax[2].grid(True)
```
